```python
# Tag: parameters
k_variant_code <- 1
k_seed_val <- 100019
k_seed_idx <- 1
```

```python
PARAM <- list()

PARAM$semilla_primigenia <- k_seed_val

# Mapeo de Variantes
Variant <- list(
    Baseline = list(name = "baseline", code = 1),
    WithoutId = list(name = "without_id", code = 2),
    WithoutMonthSnapshot = list(name = "without_month_snapshot", code = 3),
    WithoutBoth = list(name = "without_both", code = 4)
)

# Selecciono la variante según el código inyectado
PARAM$variant <- switch(k_variant_code,
  Variant$Baseline,
  Variant$WithoutId,
  Variant$WithoutMonthSnapshot,
  Variant$WithoutBoth
)

# Exp ID (usando formato de dos dígitos para la semilla: 01 a 10)
PARAM$experimento <- sprintf("E16%d%02d", PARAM$variant$code, k_seed_idx)

cat(sprintf("Configurado: Exp %s | Variante: %s | Semilla: %d\n", 
            PARAM$experimento, PARAM$variant$name, PARAM$semilla_primigenia))
```
